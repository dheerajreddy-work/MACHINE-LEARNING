## 2. Introduction to NumPy

# NumPy

**NumPy (Numerical Python)** is the fundamental package for **scientific computing in Python**. It is a Python library that provides a powerful **multidimensional array object (`ndarray`)**, various derived objects (such as **masked arrays** and **matrices**), and a wide range of routines for performing fast numerical operations.

## Features of NumPy

NumPy provides functions for:

- Mathematical operations
- Logical operations
- Shape manipulation
- Sorting and searching
- Input and Output (I/O)
- Discrete Fourier Transforms (FFT)
- Basic Linear Algebra
- Statistical operations
- Random number generation and simulation

## Characteristics (Restrictions) of NumPy Arrays

NumPy arrays have the following restrictions:

- All elements in an array must be of the **same data type** (homogeneous).
- Once an array is created, its **size cannot be changed**.
- Arrays must have a **rectangular shape** (not jagged).  
  For example, every row in a two-dimensional array must have the same number of columns.

## Advantages of These Restrictions

Because of these restrictions, NumPy arrays are:

- Faster than Python lists.
- More memory efficient.
- Optimized for numerical and scientific computations.
- Convenient for handling large datasets.

## Performance and Speed

NumPy achieves high performance through:

- **Vectorized Operations:** Performs operations on entire arrays without explicit Python loops, making computations significantly faster.
- **Implementation in C:** Many of NumPy's core functions are written in **C**, providing optimized execution.
- **Contiguous Memory Allocation:** Stores array elements in contiguous blocks of memory, improving CPU cache utilization and reducing processing overhead.

## NumPy's Main Object (`ndarray`)

The primary object in NumPy is the **homogeneous multidimensional array**, called **`ndarray`**.

An `ndarray`:

- Is a table of elements (usually numbers).
- Stores elements of the **same data type**.
- Is indexed using a **tuple of non-negative integers**.
- Can represent **1D, 2D, 3D, or higher-dimensional** data.

## 4. Array Attributes
Every ndarray exposes metadata about itself: number of dimensions (`ndim`), `shape`, total element count (`size`), element `dtype`, bytes per element (`itemsize`), total bytes (`nbytes`), the raw memory buffer (`data`), and `strides` — how many bytes to skip to move to the next element along each axis.

**Important Attributes**

```python
print(a.ndim)  # Printing the number of dimensions
print(a.shape)  # Printing the shape of the array
print(a.size)  # Printing the total number of elements in the array
print(a.dtype)  # Printing the data type of the array elements
print(a.itemsize)  # Printing the size of each element in bytes
print(a.nbytes)  # Printing the total size of the array in bytes
print(a.data)  # Printing the buffer(memory addr) containing the actual elements of the array
```

```python
import numpy as np

x = np.array([[1, 2, 3], [4, 5, 6]], dtype=np.int16)
print(x.itemsize)
print(x.strides)  # Output: (6, 2) # how many bytes one must move in memory to go to the next element along each dimension of the array
```

## 11. `flatten()` vs `ravel()`
Both collapse an array to 1D, but **`flatten()` always returns a copy** (independent memory), while **`ravel()` returns a view whenever possible** (shares memory with the original). Modifying a raveled array can therefore modify the original array too.

```python
b=a
print(b)
print(a.flatten())
print(a)
print(b)
print(a.ravel())
print(a)
print(b)
```

```python

arr = np.array([[1, 2],
                [3, 4]])
```

```python
r = arr.ravel()
r[0] = 100

print(r)
print(arr)
```

```python
arr = np.array([[1, 2],
                [3, 4]])

f = arr.flatten()
f[0] = 100

print(f)
print(arr)
```

## 15. Copies vs Views vs Assignment
Three very different behaviors when "copying" an array:

| Method | New object? | Shares data? | Changes propagate? |
|---|---|---|---|
| `b = a` (assignment) | No (same object) | Yes | Yes, both ways |
| `b = a.view()` | Yes | Yes (same buffer) | Yes, both ways |
| `b = a.copy()` | Yes | No (new buffer) | No, fully independent |

## 25. Linear Algebra (Inverse & Determinant)
`numpy.linalg.inv(A)` computes the inverse of a square matrix A, such that `A @ A_inv` gives (approximately) the identity matrix. `numpy.linalg.det(A)` computes its determinant.

```python
import numpy as np

# Define the matrix
A = np.array([[4, 7], [2, 6]])

# Compute the matrix inverse
A_inv = np.linalg.inv(A)
print(A_inv)
```

```python
import numpy as np

# Define the matrix
A = np.array([[1, 2], [3, 4]])

# Calculate the determinant
det_A = np.linalg.det(A)
print("Determinant of A:", det_A)
```

## 26. Masked Arrays

A masked array in NumPy is a special type of array that includes an additional Boolean mask, which marks certain elements as invalid or masked. This allows you to work with data that has missing or invalid values without having to modify the original data.

```python
import numpy as np
import numpy.ma as ma

# Create a normal NumPy array
data = np.array([1, 2, -999, 4, 5])

# Mask invalid values (-999 treated as missing data)
masked_data = ma.masked_equal(data, -999)

print("Original Data:", data)
print("Masked Data:", masked_data)

# Perform operations while ignoring masked values
mean_value = masked_data.mean()
print("Mean (ignoring masked values):", mean_value)
```

## 27. Interview Questions — Quick Revision (Appendix)

A rapid-fire recap of the concepts most often asked about NumPy in interviews. Each point links back to the sections above — use this as a final revision pass.


**Q1. What is NumPy?**

An open-source Python library for fast numerical computing, built around the homogeneous, multi-dimensional `ndarray` and a large collection of vectorized mathematical routines. Arrays are created with `np.array()`, `np.zeros()`, `np.ones()`, `np.full()`, `np.arange()`, or `np.linspace()` (Sections 3, 5, 6).

**Q2. Main features of NumPy**

Speed and memory efficiency from contiguous storage, a large library of mathematical functions, broadcasting, easy integration with pandas/SciPy/scikit-learn, native multi-dimensional arrays, rich indexing/slicing, and fine control over memory layout (dtype, strides).

**Q3. Dot product of two arrays**

Use `a.dot(b)`, `np.dot(a, b)`, or the `@` operator — all three compute the same matrix/vector product (Section 7).

**Q4. Shallow copy vs deep copy**

A shallow copy (`a.view()`, or plain assignment `b = a`) shares the same underlying data buffer, so changes in one are visible in the other. A deep copy (`a.copy()`) allocates independent memory, so the two arrays never affect each other (Section 15).

**Q5. Reshaping an array**

`array.reshape(new_shape)` or `np.reshape(array, new_shape)` both return a new view/copy with the given shape, as long as the total element count is unchanged (Section 12).

**Q6. Element-wise operations**

Standard operators (`+ - * /`) and `np.power()` apply automatically element-by-element when arrays share the same shape — no explicit loop required (Section 7).

**Q7. Generating random numbers**

`np.random.rand()` gives uniform floats in [0,1), `np.random.randint()` gives random integers in a range, `np.random.randn()` samples from a standard normal distribution, and `np.random.seed()` (or the modern `np.random.default_rng(seed)`) makes results reproducible (Section 8).

**Q8. Array from a Python list**

Pass the list straight into `np.array()`, e.g. `np.array([1, 2, 3, 4, 5])` (Section 3).

**Q9. Conditional (boolean) indexing**

A comparison like `arr > 3` produces a boolean array; indexing with that boolean array (`arr[arr > 3]`) returns only the elements where the condition is true (Section 18).

**Q10. Common NumPy dtypes**

int, float, complex, bool, object, and datetime, among others — dtype controls how each element is stored in memory and what operations are valid on it.

**Q11. Vertical concatenation**

`np.vstack((a, b))` or `np.concatenate((a, b), axis=0)` stack arrays row-wise (Sections 13, 17).

**Q12. Matrix inversion**

`numpy.linalg.inv(A)` returns the inverse of a square matrix A such that `A @ A_inv` is the identity matrix (Section 25).

**Q13. `var()` and `mean()`**

`np.var(a, axis=None)` computes variance (spread of the data); `np.mean(a, axis=None)` computes the arithmetic average. Both accept an `axis` to reduce along rows or columns instead of the whole array (Section 20).

**Q14. Flattening a multi-dimensional array**

`array.flatten()` returns an independent 1D copy; `array.ravel()` returns a 1D view that shares memory with the original whenever possible (Section 11).

**Q15. Identifying outliers**

Two common approaches: (a) flag values further than *k* standard deviations from the mean, or (b) use the IQR rule — flag values below `Q1 - 1.5*IQR` or above `Q3 + 1.5*IQR`, computed with `np.percentile()`.

**Q16. Removing NaN / missing values**

Build a boolean mask with `np.isnan(arr)` and index with its inverse (`arr[~mask]`), or use `np.ma.masked_invalid(arr).compressed()` to drop invalid entries via a masked array (Section 26).

**Q17. Slicing vs indexing**

Slicing (`arr[1:5]`) extracts a contiguous range using `:` and returns a view; indexing (`arr[2]`, `arr[[1,3,5]]`) picks out specific, possibly non-contiguous elements (Section 10).

**Q18. Creating an array of repeated values**

Use `np.full(shape, value)`, or broadcasting like `7 * np.ones(5)`, or a list comprehension turned into an array (Section 5).

**Q19. What is a masked array?**

A NumPy array paired with a boolean mask marking certain entries as invalid, so statistics like `.mean()` automatically skip the masked values without altering the original data (Section 26).

**Q20. Broadcasting**

NumPy's mechanism for combining arrays of different but compatible shapes without physically duplicating data — a smaller array (or scalar) is conceptually stretched to match the shape of the larger one during the operation (Section 19).

**Q21. Sorting ascending/descending**

`np.sort(arr)` sorts ascending and returns a new array. For descending order, sort ascending and reverse it (`np.sort(arr)[::-1]`), or use `arr[np.argsort(-arr)]` (Section 16).

**Q22. Why arrays beat plain Python lists**

Better performance and memory layout (contiguous, homogeneous storage), native support for vectorized math, broadcasting, and true multi-dimensional structure — all backed by compiled C code.

**Q23. `reshape()` vs `resize()`**

`reshape()` returns a new view/copy without touching the original (element count must match exactly). `resize()` modifies the array in place and can even pad with zeros or truncate data if the new size differs from the original (Section 12).

**Q24. `vstack()` vs `hstack()`**

`vstack()` stacks arrays along axis 0 (adds rows, requires matching column counts); `hstack()` stacks along axis 1 (adds columns, requires matching row counts) (Section 13).

**Q25. Eigenvalues and determinant**

`np.linalg.eigvals(matrix)` returns the eigenvalues of a matrix; `np.linalg.det(matrix)` returns its determinant (Section 25).

**Q26. Comparing two arrays**

`(arr1 == arr2).all()` checks element-wise equality and reduces it to a single boolean, while `np.array_equal(arr1, arr2)` checks both shape and values in one call.

**Q27. QR decomposition**

`numpy.linalg.qr(a, mode='reduced')` factors a matrix A into an orthogonal matrix Q and an upper-triangular matrix R such that A = Q @ R.

**Q28. What is an ndarray?**

NumPy's core data structure: a homogeneous (single dtype), potentially N-dimensional array optimized for fast mathematical and array-oriented operations (Section 4).

**Q29. Vectorization**

Performing an operation on an entire array at once (e.g. `arr ** 2`) instead of looping element by element in Python — NumPy pushes the loop down into compiled C code, which is far faster (Section 7).

**Q30. `copy()` vs `view()` vs `=`**

`b = a` creates another reference to the *same* object (fastest, zero copy). `b = a.view()` creates a new array object with independent shape/attributes but shared data. `b = a.copy()` creates a fully independent array with its own data (slowest, but safest) (Section 15).

**Q31. `shape` vs `size`**

`shape` is a tuple describing the length along each axis (e.g. `(3, 4)`); `size` is a single integer — the total element count, i.e. the product of the shape tuple (Section 4).

**Q32. Python sequences vs NumPy arrays vs pandas**

Python lists/tuples can hold mixed types and are slow for numeric work. NumPy arrays are homogeneous, n-dimensional, and fast (C-backed). pandas Series/DataFrame are built on top of NumPy but add labeled indexing, alignment, and easier handling of missing data — better suited to tabular data analysis.

**Q33. DataFrame to NumPy array**

Use the `.values` attribute (or the more modern `.to_numpy()`) on a pandas DataFrame to get the underlying NumPy array.

**Q34. Reversing an array**

Slice with a step of -1: `arr[::-1]` (Section 10).

**Q35. Why NumPy is faster than lists**

Homogeneous, contiguous memory (better CPU cache use), vectorized C-level loops instead of Python-level loops, and use of optimized linear-algebra libraries (BLAS/LAPACK) under the hood.

**Q36. Counting occurrences of integers**

`np.bincount(arr)` counts how many times each non-negative integer appears — index i of the result is the count of value i in the array.

**Q37. Max / min of an array**

`np.max(arr)` and `np.min(arr)` (or `arr.max()`/`arr.min()`) return the largest/smallest element; pass `axis=` to reduce along rows or columns instead of the whole array (Section 20).

**Q38. Slicing/indexing for data cleaning**

Boolean indexing can replace values matching a condition in place (e.g. `data[data < 0] = 0` to clip negatives), while slicing/boolean selection can filter out a clean subset (e.g. `data[data > 2]`) (Section 18).

**Q39. Finding unique elements**

`np.unique(array)` returns the sorted array of distinct values, with optional flags (`return_index`, `return_counts`, `return_inverse`) for extra metadata (Section 21).
