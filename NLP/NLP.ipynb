***Stemming***

Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words knowm as lemma.

```python
## Classification Problem
## Comments of product is a positive review or negative review
## Reviews----> eating, eat,eaten [going,gone,goes]--->go

words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

***PorterStemmer***

```python
##Porter Stemmer one of technique for stemming
from nltk.stem import PorterStemmer
stemming = PorterStemmer()
```

```python
for word in words:
  print(word,"--->",stemming.stem(word))
```

```python
stemming.stem('congratulations') ## you can see for history and congratulations , it is not giving meaningful word (disadv of stemming , we can fix using lemmatization.)
```

***RegexpStemmer class***

NLTK has RegexpStemmer class with the help of which we can easily implement Regular Expression Stemmer algorithms. It basically takes a single regular expression and removes any prefix or suffix that matches the expression. Let us see an example

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

***Lemmatization***

***Wordnet Lemmatizer***

(technique used for lemmatization)

Lemmatization technique is like stemming. The output we will get after lemmatization is called ‘lemma’, which is a root word rather than root stem, the output of stemming. After lemmatization, we will be getting a valid word that means the same thing.

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma. Let us understand it with an example −

```python
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
```

```python
'''
POS
Noun n
Verb v
Adjective a
Adverb r
'''
lemmatizer.lemmatize('going') ## by default going is noun
```

```python
lemmatizer.lemmatize('going',pos='v')
```

```python
for word in words:
  print(word,"--->",lemmatizer.lemmatize(word))
```

```python
for word in words:
  print(word,"--->",lemmatizer.lemmatize(word,pos='v'))
```

```python
lemmatizer.lemmatize('goes')
```

```python
## stemming is faster than lemmatization
## so we use this when corpus is less , ex chatbot , Q&A, text summarization
```

```python
## it is best to make your own stopwords because In this stopwords given above contains -- not,couldn't, shouldn't , for some projects like sentimental analyisis or for some other applications this not , couldn't are imp
```

```python
stemmer=PorterStemmer()
```

```python
sentences=nltk.sent_tokenize(paragraph)
```

```python
## apply stopwords and filter and then apply stemming
for i in range(len(sentences)):
  words=nltk.word_tokenize(sentences[i])
  newWords=[stemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
  sentences[i]=' '.join(newWords)
```

```python
print(sentences)
```

```python
from nltk.stem import SnowballStemmer
snowballstemmer=SnowballStemmer('english')
```

```python
## Apply Stopwords And Filter And then Apply Snowball Stemming

for i in range(len(sentences)):
    words=nltk.word_tokenize(sentences[i])
    words=[snowballstemmer.stem(word) for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
```

```python
print(sentences)
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer=WordNetLemmatizer()
```

```python
## Apply Stopwords And Filter And then Apply Snowball Stemming

for i in range(len(sentences)):
    #sentences[i]=sentences[i].lower()
    words=nltk.word_tokenize(sentences[i])
    words=[lemmatizer.lemmatize(word.lower(),pos='v') for word in words if word not in set(stopwords.words('english'))]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
```

```python
import pandas as pd
messages=pd.read_csv('SMSSpamCollection',sep='\t',names=["label","message"])
```

```python
messages
```

```python
# Data cleaning and Preprocessing
import re
import nltk
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
corpus=[]
for i in range(0,len(messages)):
  review=re.sub('[^a-zA-z]',' ',messages['message'][i])
  review=review.lower()
  review=review.split()
  review=[ps.stem(word) for word in review if not word in stopwords.words('english')]
  review=' '.join(review)
  corpus.append(review)
```

```python
corpus
```

```python
## create the bag of words
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=100)
## for binary bow binary=True
```

```python
X=cv.fit_transform(corpus).toarray()
X.shape
```

```python
## we have given 100 top feature , to find what are the features
cv.vocabulary_
```

```python
# 'go': np.int64(22),
# 'great': np.int64(25),
# this is not freq , it is the col no , if you have hello:(1)  then hello is the first feature (x1)
```

```python
## create the bag of words and n-gram
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=1000,ngram_range=(1,2))
X=cv.fit_transform(corpus).toarray()
```
