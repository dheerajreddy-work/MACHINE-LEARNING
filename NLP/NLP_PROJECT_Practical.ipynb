## ⚠️ Preventing Data Leakage

Initially, it may seem natural to perform **Bag of Words (BoW)** or **TF-IDF** on the entire dataset before splitting it into training and testing sets. However, this introduces **data leakage**.

### Why is this a problem?

BoW and TF-IDF learn the vocabulary and word statistics (such as document frequencies in TF-IDF) from the data they are fitted on. If they are fitted on the **entire dataset**, information from the **test set** is indirectly used during training.

As a result, the model gets access to information that would not be available in a real-world scenario, leading to overly optimistic evaluation results.

### ✅ Correct Workflow

1. **Preprocessing and Cleaning**
2. **Train-Test Split**
3. **Fit BoW/TF-IDF on the Training Data only**
4. **Transform both Training and Test Data using the fitted vectorizer**
5. **Train the Machine Learning Model**
6. **Evaluate on the Test Set**

```python
def avg_word2vec(doc):
    # remove out-of-vocabulary words
    sent = [word for word in doc if word in model.wv.index_to_key]
    if sent:
        return np.mean([model.wv[word] for word in sent], axis=0)
    else:
        # Return a zero vector if no in-vocabulary words are found
        return np.zeros(model.wv.vector_size)
```

***Kindle Review -- Sentimental Analysis***

```python
import pandas as pd
df=pd.read_csv('kindle_review.csv')
```

```python
df.head()
```

```python
## so many unnecessary columns i.e features
## reduce them
df=df[['reviewText','rating']]
```

```python
df.head()
```

**Preprocessing and Cleaning**

```python
## Preprocessing and Cleaning
df.isnull().sum()
```

```python
df['rating'].unique()
```

```python
df['rating'].value_counts()
```

```python
## balanced dataset
```

```python
# 1 means positive 0 means negative
df['rating']=df['rating'].apply(lambda x:1 if x>3 else 0)
```

```python
df['rating'].value_counts()
```

```python
df.head()
```

```python
## we are going to remove special chars,utls,html tags ,stopwords, remove any additional spaces
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
from bs4 import BeautifulSoup
import re
```

```python
## Removing special characters
df['reviewText']=df['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
df['reviewText']=df['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
## Remove url
df['reviewText']=df['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove html tags
df['reviewText']=df['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
## Remove any additional spaces
df['reviewText']=df['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
df.head()
```

```python
## Lemmatizer
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('wordnet')
```

```python
lemmatizer=WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
df['reviewText']=df['reviewText'].apply(lambda x:lemmatize_words(x))
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df['reviewText'],df['rating'],test_size=0.20)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
bow=CountVectorizer()
X_train_bow=bow.fit_transform(X_train).toarray()
X_test_bow=bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X_train_tfidf=tfidf.fit_transform(X_train).toarray()
X_test_tfidf=tfidf.transform(X_test).toarray()
```

```python
X_train_bow.shape
```

```python
X_train_tfidf.shape
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow=GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf=GaussianNB().fit(X_train_tfidf,y_train)
```

```python
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
y_pred_bow=nb_model_bow.predict(X_test_bow)
y_pred_tfidf=nb_model_bow.predict(X_test_tfidf)
```

```python
c=confusion_matrix(y_test,y_pred_bow)
print(c)
```

```python
print("BOW accuracy: ",accuracy_score(y_test,y_pred_bow))
```

```python
confusion_matrix(y_test,y_pred_tfidf)

```

```python
print("TFIDF accuracy: ",accuracy_score(y_test,y_pred_tfidf))
```
