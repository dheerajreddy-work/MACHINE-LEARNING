```python
def extract_episodes(txt):
    check = False
    data=""
    for i in txt:
        if i==')':
            check=False
            return data
            break
        
        if check==True:
            data+=i
        if i=='(':
            check=True
    return data
        
```

```python
def extract_time(txt):
    #after ) there are 18 char that mention time stamp
    check=False
    time_stamp=""
    for i in range(len(txt)):
        if check==True:
            for j in range(19):
                time_stamp+=txt[i+j]
            return time_stamp

        if txt[i]==')':
            check=True
```

```python
df['time_stamp']=df['Title'].apply(extract_time)
```

```python
def extract_members(txt):
    check=False
    k=0
    mem=""
    for i in range(len(txt)):
        if check==True:
            k=i+18
            while(txt[k]!=" "):
                mem+=txt[k]
                k+=1
            return mem


        if txt[i]==')':
            check=True
```

```python
from datetime import datetime

def experience_in_months(duration: str) -> int:
    """
    Example input: 'Apr 2009 - Jul 2010'
    Returns: 15
    """
    start_str, end_str = map(str.strip, duration.split("-"))

    start = datetime.strptime(start_str, "%b %Y")
    end = datetime.strptime(end_str, "%b %Y")

    return (end.year - start.year) * 12 + (end.month - start.month)
```

```python
df["months"]=df['time_stamp'].apply(experience_in_months)
```
