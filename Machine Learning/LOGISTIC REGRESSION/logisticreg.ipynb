Hyperparameter Tuning

````python
model=LogisticRegression()
penalty=['l1','l2','elasticnet']
c_values=[100,10,1.0,0.1,0.01]
solver=['newton-cg','lbfgs','liblinear','sag','saga']
````

````python
params=dict(penalty=penalty,C=c_values,solver=solver)
````

````python
from sklearn.model_selection import GridSearchCV
from sklearn.model_selection import StratifiedKFold
cv=StratifiedKFold() ## default cv=5
grid=GridSearchCV(estimator=model,param_grid=params,scoring='accuracy',cv=cv,n_jobs=-1)
````

````python
grid
````

````python
grid.fit(X_train,y_train)
````

Your error is because **GridSearchCV tries every combination** of `penalty` and `solver`, but some combinations are invalid.

For example:

```python
penalty='l1'
solver='newton-cg'
```

❌ Not allowed.

### Valid combinations

| Solver      | Penalty                  |
| ----------- | ------------------------ |
| `newton-cg` | `l2`                     |
| `lbfgs`     | `l2`                     |
| `liblinear` | `l1`, `l2`               |
| `sag`       | `l2`                     |
| `saga`      | `l1`, `l2`, `elasticnet` |

So instead of:

```python
params = dict(
    penalty=penalty,
    C=c_values,
    solver=solver
)
```

use:

```python
params = [
    {'penalty':['l2'], 'C':c_values,
     'solver':['newton-cg','lbfgs','sag']},

    {'penalty':['l1','l2'], 'C':c_values,
     'solver':['liblinear']},

    {'penalty':['l1','l2','elasticnet'], 'C':c_values,
     'solver':['saga'], 'l1_ratio':[0.5]}
]
```

The `nan` values simply mean **that parameter combination failed**.

So, **more clusters = more separate groups within each class**.

---

### 2. `n_redundant`

A **redundant feature** is a feature that contains information already available in other features.

Example:

```text
Feature 1 = Age
Feature 2 = Salary
Feature 3 = 2 × Age + some relationship
```

Feature 3 is redundant because it doesn't provide much new information.

```python
n_redundant=0
```

### Easy way to remember

* **cluster** → How many **groups** inside each class?
* **redundant** → How many **duplicate/repeated-information features**?

So your code creates **10,000 samples, 2 features, 1 cluster per class, and 0 redundant features**, with `99%` of samples intended for one class.

If you mean **ROC curve for Logistic Regression thresholds**, the idea is:

### 1. Logistic Regression gives probabilities

Example:

| Actual | Predicted probability |
| ------ | --------------------: |
| 1      |                  0.90 |
| 1      |                  0.70 |
| 0      |                  0.60 |
| 0      |                  0.30 |
| 1      |                  0.20 |

Normally, we use **threshold = 0.5**:

* probability ≥ 0.5 → predict **1**
* probability < 0.5 → predict **0**

### 2. Change the threshold

ROC checks many thresholds:

**Threshold = 0.9**

**Threshold = 0.7**

**Threshold = 0.5**

**Threshold = 0.3**

**Threshold = 0.1**

For every threshold, calculate:

* **TPR (True Positive Rate)** = TP / (TP + FN)
* **FPR (False Positive Rate)** = FP / (FP + TN)

Then plot:

**X-axis = FPR**
**Y-axis = TPR**

### Key point

**ROC is basically checking how the model performs when we move the classification threshold from 0 → 1.**

So:

```text
Logistic Regression
       ↓
Predicted probabilities
       ↓
Try different thresholds
       ↓
TPR and FPR for each threshold
       ↓
ROC Curve
       ↓
AUC
```

**AUC** summarizes the ROC curve:

* AUC ≈ 1 → excellent
* AUC ≈ 0.5 → random guessing
* AUC < 0.5 → worse than random
